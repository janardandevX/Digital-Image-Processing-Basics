```python
import cv2
import matplotlib.pyplot as plt

img = cv2.resize(cv2.imread("D:\\Docs\\photo.jpeg"), (600,720))

rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
yuv = cv2.cvtColor(img, cv2.COLOR_BGR2YUV)
hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
ycbcr = cv2.cvtColor(img, cv2.COLOR_BGR2YCrCb)
cmy = 255 - rgb
rot = cv2.rotate(img, cv2.ROTATE_180)

images = [rgb, gray, yuv, hsv, cmy, ycbcr, cv2.cvtColor(rot, cv2.COLOR_BGR2RGB), rgb]
titles = ["RGB", "Grayscale", "YUV", "HSV", "CMY", "YCbCr", "Rotated 180°", "Original"]

plt.figure(figsize=(12, 8))
for i, (image, title) in enumerate(zip(images, titles)):
    plt.subplot(2, 4, i+1)
    plt.imshow(image, cmap="gray" if title == "Grayscale" else None)
    plt.title(title)
    plt.axis("off")

plt.tight_layout()
plt.show()
```
