### Min Max Stretching

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Read the image in grayscale

img = cv2.imread(r"C:\Users\janar\Digital Image Processing\Images\girl.jpg", 0)
```

```python
#Find the minimum and maximum pixel values in the original image

r_min = np.min(img)
r_max = np.max(img)
```

```python
stretched_img = ((img.astype(np.float32) - r_min) / (r_max - r_min)) * 255
```

```python
stretched_img = np.clip(stretched_img, 0, 255).astype(np.uint8)
```

```python
hist_original = cv2.calcHist([img], [0], None, [256], [0, 256])
hist_stretched = cv2.calcHist([stretched_img], [0], None, [256], [0,256])
```

```python
#Display the results

plt.figure(figsize=(12, 8))

plt.subplot(3, 2, 1)
plt.imshow(img, cmap='grey', vmin=0, vmax=255)
plt.title(f'Original (Min: {r_min}, Max: {r_max})')
plt.axis('off')

plt.subplot(3, 2, 2)
plt.plot(hist_original, color='black')
plt.title('Original Histogram')
plt.xlim([0, 256])

plt.subplot(3, 2, 3)
plt.imshow(stretched_img, cmap='gray', vmin=0, vmax=255)
plt.title('Linearly Stretched (Min: 0, Max: 255)')
plt.axis('off')

plt.subplot(3, 2, 4)
plt.plot(hist_stretched, color='black')
plt.title('Stretched Histogram')
plt.xlim([0, 256])

plt.tight_layout()
plt.show()
```
