### Image Averaging

```python
import cv2 
img1 = cv2.imread("D:\\Docs\\photo.jpeg") 
img2 = cv2.imread("D:\\Docs\\photo.jpeg")

result = cv2.addWeighted( img1, 0.5, img2, 0.5, 0 ) 

cv2.imshow("Image1: ", img1) 
cv2.imshow("Image2: ", img2) 
cv2.imshow("Average", result) 
cv2.waitKey(0) 
cv2.destroyAllWindows()
```
