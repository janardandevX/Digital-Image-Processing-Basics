### Image Addition

```python
import cv2

img1 = cv2.imread("D:\\Docs\\photo.jpeg")
img2 = cv2.imread("D:\\Docs\\photo.jpeg")

result = cv2.add(img1, img2)

img1 = cv2.resize(img1, (600, 720))
img2 = cv2.resize(img2, (600, 720))
result = cv2.resize(result, (600, 720))

cv2.imshow("Image1: ", img1)
cv2.imshow("Image2: ", img2)
cv2.imshow("Addition", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
