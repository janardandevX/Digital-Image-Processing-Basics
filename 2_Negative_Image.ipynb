```python
import cv2
```

```python
#Read grayscale image
img = cv2.imread("D:\\Docs\\photo.jpeg")
img = cv2.resize(img, (600, 700))

#Create the negative
negative = 255 - img
```

```python
#Display results

cv2.imshow("Original:", img)
cv2.imshow("Negative:", negative)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
